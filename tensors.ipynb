```python
import torch
import numpy as np
```

## Tensor Initialization

```python
data = [[1, 2], [3, 4]]
x_data = torch.tensor(data)
x_data
```

```python
np_data = np.array(data)
x_np = torch.from_numpy(np_data)
x_np
```

```python
x_ones = torch.ones_like(x_data)
print(f"Ones: {x_ones}")

x_rand = torch.rand_like(x_data, dtype=torch.float)
print(f"Rand: {x_rand}")
```

```python
shape = (3, 4)

rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: {rand_tensor}")
print(f"Ones Tensor: {ones_tensor}")
print(f"Zeros Tensor: {zeros_tensor}")
```

## Tensor Attributes

```python
tensor = torch.rand(3, 4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device of tensor: {tensor.device}")
```

```python
device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else 'cpu'
tensor = tensor.to(device)
print(f"Device of tensor: {tensor.device}")
```

## Tensor Operations

```python
t = torch.ones(4, 4)
t[2, :] = 0
t
```

```python
t1s = torch.ones(4, 4)
trs = torch.rand(4, 4)
t = torch.cat([t1s, trs])
t
```

```python
t = torch.stack([t1s, trs])
t
```

```python
t1 = torch.ones(3, 4)
t2 = torch.rand(3, 4)

print(f"Elementwise multiplication: {t1 * t2}")
print(f"Matrix multiplication: {t1 @ t2.T}")
```
